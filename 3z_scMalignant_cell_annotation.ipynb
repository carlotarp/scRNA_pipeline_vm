### Needs to be executed in conda scmalignant environment.

```python
# Import Libraries
from scMalignantFinder import classifier
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import scanpy as sc
```

```python
# Set Paths
project_path = "/home/usuario/PROJECTS/260724_victor_scRNA/"
results_path = project_path + "results/"
results_GEMX_CL_path = results_path + "GEMX/Clustering/"
pretrain_path = project_path + "data/scMalignantdata"
```

```python
# Load scMalignantFinder Classifier
model = classifier.scMalignantFinder(
    test_input= results_GEMX_CL_path+"clustered_normalized_data.h5ad",
    pretrain_dir= pretrain_path,
    norm_type=True,
    n_thread=1)
model.load()
```

```python
# Predict Data
pred = model.predict()
cell_metadata = pred.obs.copy()
```

```python
# Export Predictions
cell_metadata.to_csv(results_GEMX_CL_path+'scmalignantfinder_metadata.tsv', sep='\t')
```

```python
# Cargar el AnnData exportado desde R (con PCA/UMAP ya calculados)
adata = sc.read_h5ad("/home/usuario/PROJECTS/260724_victor_scRNA/results/GEMX/Clustering/clustered_normalized_data.h5ad")
```

```python
adata.obs["scMalignantFinder_prediction"] = cell_metadata.loc[adata.obs_names, "scMalignantFinder_prediction"]
adata.obs["malignancy_probability"] = cell_metadata.loc[adata.obs_names, "malignancy_probability"]
```

```python
sc.settings.figdir = results_GEMX_CL_path
```

```python
# Visualizar PCA y UMAP
sc.pl.pca(adata, color="scMalignantFinder_prediction", show=False, save="_scMalignantFinder.png")
sc.pl.umap(adata, color="scMalignantFinder_prediction", show=False, save="_scMalignantFinder.png")
sc.pl.umap(adata, color="malignancy_probability", cmap="viridis", show=False, save="_scMalignantFinder_Prob.png")
```
